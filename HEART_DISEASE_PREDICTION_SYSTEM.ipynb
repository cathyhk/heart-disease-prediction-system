##  Initial Data Inspection

The combined heart disease dataset contains 920 records and 15 columns. It includes data from four sources: Cleveland, Hungarian, Switzerland and VA.

The target variable contains 509 records labelled 1 and 411 records labelled 0.
The initial inspection has identified missing values in several columns. The largest amounts of missing data were found in `ca` (611), `thal` (486) and `slope` (309).
Two pairs of duplicate rows have been identified using pandas.


```python
# Calculate the percentage of missing values
missing_percentage = (df.isnull().sum() / len(df)) * 100
# Create a summary table
missing_summary = pd.DataFrame({
    'Missing Values': df.isnull().sum(),
    'Percentage (%)': missing_percentage.round(2)})
# Display the results
missing_summary.sort_values(
    by='Percentage (%)', ascending=False)
```

## 6. Missing Data Analysis

The missing-value analysis showed that the dataset has different levels of missing data across its features.

The `ca` column has 611 missing values (66.41%), followed by `thal` with 486 (52.83%) and `slope` with 309 (33.59%). Other columns have lower levels of missing data while `age`, `sex`, `cp`, `target`and `source` have no missing values.


```python
# Count missing values in each row
df['missing_count'] = df.isnull().sum(axis=1)
# Summarize how many missing values each row has
print("Missing values per row:")
print(df['missing_count'].value_counts().sort_index())
```

```python
# Compare missingness across sources
print("\nAverage missing values per record by source:")
print(df.groupby('source')['missing_count'].mean().round(2))

# Remove the temporary inspection column
df.drop(columns=['missing_count'], inplace=True)
```

Missing Data Patterns

The row level analysis showed that 299 records contain no missing values while 621 records have at least one missing value.
The average number of missing values per record differs across the four sources. Cleveland has the lowest average at 0.02 while VA has the highest at 3.49. Hungarian and Switzerland have averages of 2.66 and 2.22, respectively.

```python

# Display duplicate records for review
duplicates = df[df.duplicated(keep=False)].sort_values(
    by=['source', 'age']
)

display(duplicates)

# Count duplicate rows within each source
print("Duplicate rows by source:")
print(
    df[df.duplicated(keep=False)]
    .groupby('source')
    .size()
)
```

```python
#remobve the duplicate rows
# Check the dataset size before removing duplicates
print("Before:", df.shape)

# Remove duplicate rows, keeping the first occurrence
df = df.drop_duplicates()

# Check the dataset size after removing duplicates
print("After:", df.shape)

# Confirm that no exact duplicate rows remain
print("Duplicates remaining:", df.duplicated().sum())
```
